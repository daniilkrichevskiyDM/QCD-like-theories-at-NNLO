```python
#function which extracts data for different betas 
def extract_rows_by_value(matrix):
    """
    The function takes a 2D NumPy array and extracts the rows where the first column contains the values 6.9, 7.05, or 7.2.
    """
    values_to_extract = {6.9, 7.05, 7.2}  # Устанавливаем нужные значения

    # Извлекаем строки, соответствующие указанным значениям
    extracted_6_9 = matrix[np.isclose(matrix[:, 0], 6.9)]
    extracted_7_05 = matrix[np.isclose(matrix[:, 0], 7.05)]
    extracted_7_2 = matrix[np.isclose(matrix[:, 0], 7.2)]    

    return extracted_6_9, extracted_7_05, extracted_7_2

#function which extracts the data from the matrices 
def process_matrices_for_different_betas(matrices):
   
    results = []
    
    for matrix in matrices:
        mPiA = matrix[:, 3]  # mass of type A
        DmPiA = matrix[:, 4]  # error of mass of type A
        mPiASquared = mPiA * mPiA  # Squared mass of type A
        DmPiASquared = 2 * mPiA * DmPiA  # error of squared mass of type A

        mPiC = matrix[:, 5]  # mass of type C
        DmPiC = matrix[:, 6]  # error of mass of type C
        mPiCSquared = mPiC * mPiC  # Squared mass of type C
        DmPiCSquared = 2 * mPiC * DmPiC  # error of squared mass of type C

        fA = matrix[:, 7]  # decay constant of type A
        DfA = matrix[:, 8]  # error of decay constant of type A

        fC = matrix[:, 9]  # decay constant of type C
        DfC = matrix[:, 10]  # error of decay constant of type C

        #Averaging
        fAverage = (fA+fC)/2
        
        DfAverage = (DfA+DfC)/2 #statistical error
        systematic_F_error = np.abs(fA - fC)/2 #systematic error

        Fsigma_eff = np.sqrt(DfAverage**2 + systematic_F_error**2)

    
    
        # Compute x2A and x2C
        x2A = mPiASquared / (fA ** 2)
        Dx2A = 2 * (mPiASquared * fA * DfA + fA ** 2 * mPiA * DmPiA) / fA ** 4

        x2C = mPiCSquared / (fA ** 2)
        Dx2C = 2 * (mPiCSquared * fA * DfA + fA ** 2 * mPiC * DmPiA) / fA ** 4

        # Another option taking fC into account
        x2C_new = mPiCSquared / (fC ** 2)
        Dx2C_new = 2 * (mPiCSquared * fC * DfC + fC ** 2 * mPiC * DmPiC) / fC ** 4

        # X-axis
        mAverage = matrix[:, 11]  # average PCAC mass
        DmAverage = matrix[:, 12]  # error of average PCAC mass

        mu = mAverage[-1]  # mu stays fixed
        Dmu = DmAverage[-1]  # error of mu

        r = -1 + 2 * mAverage / mu  # ratio r=md/mu
        Dr = 2 / (mu ** 2) * (mAverage * Dmu + mu * DmAverage)  # error of r=mu/md

        #bared quark masses
        mu0 = matrix[:, 1]
        md0 = matrix[:, 2]

         # Store results as dictionary
        results.append({
            "mu0": mu0,
            "md0": md0,
            "mPiASquared": mPiASquared,
            "DmPiASquared": DmPiASquared,
            "mPiCSquared": mPiCSquared,
            "DmPiCSquared": DmPiCSquared,
            "fA": fA,
            "DfA": DfA,
            "fC": fC,
            "DfC": DfC,
            "fAverage": fAverage,
            "DfAverage": DfAverage,
            "DfAverage_eff": Fsigma_eff,
            "x2A": x2A,
            "Dx2A": Dx2A,
            "x2C": x2C,
            "Dx2C": Dx2C,
            "x2C_new": x2C_new,
            "Dx2C_new": Dx2C_new,
            "r": r,
            "Dr": Dr
        })
    
    return results
```

## Define the functions for predictions

```python
#ChPT functions 

def mSquaredLO(r, B0_mu):
    return B0_mu*(1+r)

def mASquared(r, B0_mu, F, L1, L3): #function with index 0
    MSquared = mSquaredLO(r, B0_mu) #squared LO mass
    return MSquared  * (1 + (1 / (64 * np.pi**2)) * (MSquared / F**2) * (-4 * L1 + 5*L3))

def mCSquared(r, B0_mu, F, L1, L3, L4):   #function with index 1 
    MSquared = mSquaredLO(r, B0_mu) #squared LO mass
    return MSquared * (1 + (1 / (64 * np.pi**2)) * (MSquared / F**2) * (-4 * L1 +  5*L3 + 8 * L4 * ((1 - r)**2 / (1 + r)**2)))

def FA(r, B0_mu, F, L1): #function with index 2
    MSquared = mSquaredLO(r, B0_mu) #squared LO mass
    return F * (1 + (1 / (32 * np.pi**2)) * (MSquared / F**2) * L1)

def a0_mPi(x, L1, L2, L3):
    x2 = x**2
    Bijnens = - (1 / (32 * np.pi)) * x2 - (1 / (6144 * np.pi**3)) * x2**2 * (3 + 24 * L1 + 16 * L2 - 15 * L3)
    return -Bijnens#-2*Bijnens 


#Derivatives

def dmASquared_dr(r, B0_mu, F, L1, L3):
    return (mASquared(r+10e-7, B0_mu, F, L1, L3)-mASquared(r, B0_mu, F, L1, L3))/10e-7

def dmCSquared_dr(r, B0_mu, F, L1, L3, L4): 
    return (mCSquared(r+10e-7, B0_mu, F, L1, L3, L4)-mCSquared(r, B0_mu, F, L1, L3, L4))/10e-7

def dFA_dr(r, B0_mu, F, L1):
    return (FA(r+10e-7, B0_mu, F, L1)-FA(r, B0_mu, F, L1))/10e-7

def da0_mPi_dx(x, L1, L2, L3):
    return (a0_mPi(x+10e-7, L1, L2, L3)-a0_mPi(x, L1, L2, L3))/10e-7
```

## Define componens of Bayes formula

```python
def log_prior(params, x_data_Kulkarni, x_err_Kulkarni, x_data_Maas, x_err_Maas):
    # Извлекаем фиксированные параметры
    L1, L2, L3, L4, B0_mu, F = params[:6]

    # Затем извлекаем r1, r2, ..., rn
    r_array = params[6:6 + n_r_point]

    # Остаток — это mpi_fpi
    mpi_fpi = params[6 + n_r_point:]

    # Проверка диапазона для B0_mu и F
    if not (0 <= B0_mu <= 0.1):
        return -np.inf
    if not (0 <= F <= 0.1):
        return -np.inf

    sigma = 10 #sigma_L in my notes
    lp = 0.0

    # Нормальный приор для L1...L4
    for L in [L1, L2, L3, L4]:
        lp += -0.5 * (L**2 / sigma**2)

    # Приор для r_i из Kulkarni
    for i in range(n_r_point):
        r_true = r_array[i]
        if r_true <= 0:
            return -np.inf
        mu_r = x_data_Kulkarni[0, i]
        sigma_r = x_err_Kulkarni[0, i]
        lp += -0.5 * ((r_true - mu_r)**2) / sigma_r**2

    # Приор для mpi_fpi из Maas
    for i in range(len(mpi_fpi)):
        mpi_fpi_true = mpi_fpi[i]
        if mpi_fpi_true <= 0:   # or < 0 if you want to allow exactly zero
            return -np.inf
        mu_mpi_fpi = x_data_Maas[i]
        sigma_mpi_fpi = x_err_Maas[i]
        lp += -0.5 * ((mpi_fpi_true - mu_mpi_fpi)**2) / sigma_mpi_fpi**2

    return lp
```

```python
#log-likelyhood

def likelihood_Kulkarni_mASquared_one_point(params, r_parameter, p, sigma_y):
    L1, L2, L3, L4, B0_mu, F = params[:6]
    return -0.5*(p - mASquared(r_parameter, B0_mu, F, L1, L3))**2/(sigma_y**2)
       
def likelihood_Kulkarni_mCSquared_one_point(params, r_parameter , p, sigma_y):
    L1, L2, L3, L4, B0_mu, F = params[:6]
    return -0.5*(p - mCSquared(r_parameter, B0_mu, F, L1, L3, L4))**2/(sigma_y**2)

def likelihood_Kulkarni_FA_one_point(params, r_parameter , p, sigma_y):#for fA we now take the average
    L1, L2, L3, L4, B0_mu, F = params[:6]
    return -0.5*(p - FA(r_parameter, B0_mu, F, L1))**2/(sigma_y**2)  

def likelihood_Maas_one_point(params, x_parameter, p, sigma_y):
    L1, L2, L3, L4, B0_mu, F = params[:6]
    return -0.5*(p - a0_mPi(x_parameter, L1, L2, L3))**2/sigma_y**2


def log_likelihood(params, y_data_Kulkarni, y_err_Kulkarni, y_data_Maas, y_err_Maas):
    # Извлекаем фиксированные параметры
    L1, L2, L3, L4, B0_mu, F = params[:6]

    # Затем извлекаем r1, r2, ..., rn по количеству n_r_point
    r_array = params[6:6 + n_r_point]

    # Остаток — это mpi_fpi
    mpi_fpi = params[6 + n_r_point:]

    likelihood = 0

    # Добавляем вклады Maas
    for i1 in range(len(mpi_fpi)):
        likelihood += likelihood_Maas_one_point(params, mpi_fpi[i1], y_data_Maas[i1], y_err_Maas[i1])

    # Добавляем вклады Kulkarni
    for i in range(n_r_point):
        likelihood += likelihood_Kulkarni_mASquared_one_point(params, r_array[i], y_data_Kulkarni[0, i], y_err_Kulkarni[0, i])
        likelihood += likelihood_Kulkarni_mCSquared_one_point(params, r_array[i], y_data_Kulkarni[1, i], y_err_Kulkarni[1, i])
        likelihood += likelihood_Kulkarni_FA_one_point(params, r_array[i], y_data_Kulkarni[2, i], y_err_Kulkarni[2, i])

    return likelihood
```

```python
def log_posterior(params, x_data_Kulkarni, y_data_Kulkarni, x_err_Kulkarni, y_err_Kulkarni,
                  x_data_Maas, y_data_Maas, x_err_Maas, y_err_Maas):

    lp = log_prior(params, x_data_Kulkarni, x_err_Kulkarni, x_data_Maas, x_err_Maas)
    if not np.isfinite(lp):  
        return -np.inf

    ll = log_likelihood(params, y_data_Kulkarni, y_err_Kulkarni, y_data_Maas, y_err_Maas)
    
    return lp + ll
```

## Run MCMC

```python
# MCMC
ndim = 6 + n_r_point  + len(Maas_x)  # number of parameters 6 parametrs of the theory, 6 uncertain positions of r, len(Maas_x) positions of mpi/fpi
nwalkers = 70  # number of chains

#L1, L2, L3, L4, B0_mu, F, r1, r2, r3, r4, r5, r6,  mpi_fpi1 ...  = params

# Initial points for chains 
initial_guess = [0, 0, 0, 0, 0.01, 0.01] + list(Kulkarni_x[0,:])  + list(Maas_x)
p0 = [
    initial_guess + np.array(
        [1e-2] * 4 + [1e-3] * 2 + [1e-2] * n_r_point  +[1e-2] * len(Maas_x)
    ) * np.random.randn(ndim)
    for _ in range(nwalkers)
]

# log-posterior for MCMC
def log_posterior_mcmc(params):
    return log_posterior(params, Kulkarni_x, Kulkarni_y, Kulkarni_dx, Kulkarni_dy,
                         Maas_x, Maas_y, Maas_dx ,Maas_dy)

# Setting up the MCMC sampler
sampler = emcee.EnsembleSampler(nwalkers, ndim, log_posterior_mcmc)

# Parameters for Burn-in and Main MCMC
burn_in_steps = 2000  # number of steps for burn-in
main_steps = 25000    # number of steps for the main sampling

# Running Burn-in
print("Running Burn-in...")
pos, prob, state = sampler.run_mcmc(p0, burn_in_steps, progress=True)
sampler.reset()  # Reset the sampler state after burn-in

# Running Main MCMC
print("Running Main MCMC...")
sampler.run_mcmc(pos, main_steps, progress=True)

# Obtaining results
samples = sampler.get_chain(flat=True)

#choose first 6 parameters
selected_samples = samples[:, :6]

# Вычисляем медианы и 1σ ошибки
medians = np.median(selected_samples, axis=0)
lower_errors = medians - np.percentile(selected_samples, 16, axis=0)
upper_errors = np.percentile(selected_samples, 84, axis=0) - medians

# Параметры
param_names = ["L1", "L2", "L3", "L4", "B0_mu", "F"]

# Вывод результатов
for name, median, lower, upper in zip(param_names, medians, lower_errors, upper_errors):
    print(f"{name}: median = {median:.6f}, -1σ = {lower:.6f}, +1σ = {upper:.6f}")

mcmc_results = {
    name: {"median": median, "-1σ": lower, "+1σ": upper}
    for name, median, lower, upper in zip(param_names, medians, lower_errors, upper_errors)
}


# Generate the corner plot for the first 6 parameters
fig = corner.corner(
    selected_samples,
    labels=param_names,
    show_titles=True,
    title_fmt=".6f"
)

# Display the plot
plt.show()
```

```python
# Load data (solution of the Bolzmann equation)
dataNc4 = np.loadtxt("/Users/daniilkrichevskiy/Desktop/Project SIMP at NLO/Fit the LECs/Plots_for_mpi_f_pi/m_f_Nc=4.csv", delimiter=",")  # Укажите правильный путь

# Extract x and y values for interpolation
x = dataNc4[:, 0]  # m
y = dataNc4[:, 1]  # m/f

# Apply log transformation to y for better interpolation in log scale
log_y = np.log(y)

# Create interpolation function in log scale
log_interp = interp1d(x, log_y, kind='linear', fill_value="extrapolate")
y_smooth = np.exp(log_interp(x))  # Convert back to linear scale

# Generate fine x values for smooth curve
x_fine = np.logspace(np.log10(x.min()), np.log10(x.max()), 300)
y_fine = np.exp(log_interp(x_fine))

def myNc4(x_new): #this function I use for mpi/fpi
    return np.exp(log_interp(x_new))

# Plot data points and interpolation
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='red', label="Data Points")
plt.plot(x_fine, y_fine, color='blue', linestyle='-', label="Log-Interpolated Function")
plt.xscale("log")  # Set x-axis to log scale
plt.xlabel("m")
plt.ylabel("m/f")
plt.title("Logarithmic Interpolation of Data Points")
plt.legend()
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.show()
```

```python
#formulas for sigma/m
def self_scattering_crossection(m, LECs = [0,0,0,0], NLO = True):
    L1_Tilde = LECs[0]
    L2_Tilde = LECs[1]
    L3_Tilde = LECs[2]

    
    f = m/myNc4(m)
    if NLO == True:
        return (1.22221e-6 * m / f**4) + (3.2937e-8 * m**3 / f**6) + (1.30716e-8 * L1_Tilde * m**3 / f**6) - (3.21056e-9 * L2_Tilde/4 * m**3 / f**6) + (3.0099e-9 * L3_Tilde * m**3 / f**6)

    if NLO == False:
        return (1.22221e-6 * m) / (f**4)


def self_scattering_crossection_with_x(x, m, LECs = [0,0,0,0], NLO = True): #here x = m/f
    L1_Tilde = LECs[0]
    L2_Tilde = LECs[1]
    L3_Tilde = LECs[2]
    
    if NLO == True:
        return ((1.22221e-6 * x**4) + (3.2937e-8 * x**6) + (1.30716e-8 * L1_Tilde * x**6)  - (3.21056e-9 * L2_Tilde/4 * x**6) + (3.0099e-9 * L3_Tilde * x**6) )/m**3
    if NLO == False:
        return (1.22221e-6 * x**4) / (m**3)

def self_scattering_crossection_NNLO(x, m, mu, LECsNLO = [0,0,0,0], LECsNNLO = [0,0,0,0,0,0,0,0,0] ): #here x = m/f
    log_term = np.log((m**2) / (mu**2))
    log_term2 = np.log((m**2) / (mu**2))**2
    prefactor = 1 / m**3 * 0.000146014 * x**8

    L0r = LECsNNLO[0]
    L1r = LECsNNLO[1]
    L2r = LECsNNLO[2]
    L3r = LECsNNLO[3]
    L4r = LECsNNLO[4]
    L5r = LECsNNLO[5]
    L6r = LECsNNLO[6]
    L7r = LECsNNLO[7]
    L8r = LECsNNLO[8]

    poly = (
    1.48757e-6
    + 1.0 * L0r**2
    + 0.0125521 * L1r + 16.0 * L1r**2
    + 0.011949 * L2r + 32.0 * L1r * L2r + 16.0 * L2r**2
    + 0.00166795 * L3r + 8.0 * L1r * L3r + 8.0 * L2r * L3r + 1.0 * L3r**2
    + 0.212514 * L4r - 10.6667 * L1r * L4r - 10.6667 * L2r * L4r - 2.66667 * L3r * L4r + 14.4762 * L4r**2
    + 0.00298724 * L5r - 2.66667 * L1r * L5r - 2.66667 * L2r * L5r - 0.666667 * L3r * L5r + 9.71429 * L4r * L5r + 1.52381 * L5r**2
    + 0.0086319 * L6r + 32.0 * L1r * L6r + 32.0 * L2r * L6r + 8.0 * L3r * L6r - 28.9524 * L4r * L6r - 12.1905 * L5r * L6r + 34.2857 * L6r**2
    + 0.002648 * L8r + 8.0 * L1r * L8r + 8.0 * L2r * L8r + 2.0 * L3r * L8r - 2.28571 * L4r * L8r - 1.80952 * L5r * L8r + 7.2381 * L6r * L8r - 0.333333 * L8r**2
    + L0r * (0.00332649 + 8.0 * L1r + 8.0 * L2r + 2.0 * L3r - 2.66667 * L4r - 0.666667 * L5r + 8.0 * L6r + 2.0 * L8r))
    log_linear = (
    -7.60823e-6 + 0.00360919 * L0r - 0.0168492 * L1r - 0.0160576 * L2r - 0.00396728 * L3r + 0.00154545 * L4r - 0.00368458 * L5r - 0.0213724 * L6r
    - 0.00384478 * L7r - 0.0000282704 * L8r) * log_term
    log_squared = 5.77494e-6 * log_term2
    resultNNLO = prefactor * (poly + log_linear + log_squared)
    
    
    return self_scattering_crossection_with_x(x, m, LECsNLO, True)+resultNNLO
```

```python
# Define the m_pi-axis range
m_pi_values = np.logspace(-3, 0, 500)  # From 10^(-3) to 10^0, log scale
num_samples = 2000

# Sample from the MCMC posterior
indices = np.random.choice(len(samples), num_samples, replace=False)
posterior_samples = samples[indices, :]

# Compute predictions for each sample
y3_samples = []
for params in posterior_samples:
    L1, L2, L3, L4 = params[:4]  # Extract relevant parameters
    y3_samples.append([self_scattering_crossection(m, LECs=[L1, L2, L3, L4], NLO=True) for m in m_pi_values])

# Convert to numpy array
y3_samples = np.array(y3_samples)

# Compute median and percentiles
y3_median = np.median(y3_samples, axis=0)
y3_lower = np.percentile(y3_samples, 0.15, axis=0)  # 3σ lower bound
y3_upper = np.percentile(y3_samples, 99.85, axis=0)  # 3σ upper bound

# Compute LO curve
y1_values = myNc4(m_pi_values)
y2_values = self_scattering_crossection(m_pi_values, LECs=[0, 0, 0, 0], NLO=False)  # LO curve
```

```python
def find_trust_region(threshold=1.0, x_max=14, step=0.01):
    trusted_x = None
    x = 0.0
    while x <= x_max:
        lo = self_scattering_crossection_with_x(x, 0.3, LECs=[mcmc_results['L1']['median'], mcmc_results['L2']['median'], mcmc_results['L3']['median'], mcmc_results['L4']['median']], NLO=False)
        nlo = self_scattering_crossection_with_x(x, 0.3, LECs=[mcmc_results['L1']['median'], mcmc_results['L2']['median'], mcmc_results['L3']['median'], mcmc_results['L4']['median']], NLO=True)
        
        if lo == 0:
            x += step
            continue
        
        relative_diff = abs(nlo - lo) / abs(lo)
        if relative_diff > threshold:
            break
        
        trusted_x = x
        x += step

    return trusted_x

find_trust_region()
```

```python
def find_trust_region(threshold=0.2, x_max=14, step=0.01):
    trusted_x = None
    x = 0.0
    while x <= x_max:
        lo = self_scattering_crossection_with_x(x, 0.3, LECs=[0, 0, 0, 0], NLO=False)
        nlo = self_scattering_crossection_with_x(x, 0.3, LECs=[0, 0, 0, 0], NLO=True)
        
        if lo == 0:
            x += step
            continue
        
        relative_diff = abs(nlo - lo) / abs(lo)
        if relative_diff > threshold:
            break
        
        trusted_x = x
        x += step

    return trusted_x

find_trust_region()
```
